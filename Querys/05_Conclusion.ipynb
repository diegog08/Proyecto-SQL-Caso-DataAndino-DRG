#Conclusión de Negocio
Le pagamos USD 60,228 a las agencias por reservas que después se cancelaron, y la mayoría de esas cancelaciones pasan en el City Hotel. Con las hipótesis vi que los hijos casi no cambian nada, pero el huésped que pide algo especial o modifica su reserva cancela muy poco, mientras que el que reserva y desaparece cancela casi la mitad de las veces. Yo renegociaría con las agencias para pagar solo cuando la reserva se queda, y les seguiría la pista a esas reservas donde el cliente nunca dio señales de vida.
